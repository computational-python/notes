## collaboration workflow

* contribute to an open source project
* local development branch with changes
* push changes to your own github account with a copy (fork)
* pull request: ask the maintainer to consider your changes for merging

```python
!git branch development
```

```python
!git log --oneline --all
```

```python
# activate the new branch
!git switch development
```

```python
!git log --oneline --all
```
