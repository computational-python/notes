## Branching

```python
n = 0
if n > 0:
    print(f"{n} is positive")
elif n < 0:
    print(f"{n} is negative")
else:
    print(f"{n} must be zero")
```

```python
# variation without elif
n = 0
if n > 0:
    print(f"{n} is positive")
else:
    if n < 0:
        print(f"{n} is negative")
    else:
        print(f"{n} must be zero")
```

```python
# data in if statements convert to True/False
if 1:
    print("1 is True")
```

```python
# strings
#string = "abc"
string = ""
if string:
    print("string is", string)
else:
    print("string is empty")
```

```python
# lists
#l = [1, 2, 3]
l = []
if l:
    print("list is", l)
else:
    print("list is empty")
```

```python
# dicts
#d = {'a': 1}
d = {}
if d:
    print("dict is", d)
else:
    print("dict is empty")
```

## Example
Number guessing game

```python
secret = 11

guess = input("Guess a number: ")
guess = int(guess)

while True:
    if guess > secret:
        print("Too large")
    elif guess < secret:
        print("Too small")
    else:
        print("You win")
        break  # stop looping continue with next lines

    guess = input("Guess a number: ")
    guess = int(guess)

print("Game over")
```

retrict number of guesses

```python
secret = 11
max_guesses = 5

for _ in range(max_guesses):
    guess = input("Guess a number: ")
    guess = int(guess)
    
    if guess > secret:
        print("Too large")
    elif guess < secret:
        print("Too small")
    else:
        print("You win")
        break
else: # executed in the absence of break
    print("You lose")


print("Game over")
```

get the secret from another file (module)

```python
import settings
secret = settings.secret_number

max_guesses = settings.max_number

for _ in range(max_guesses):
    guess = input("Guess a number: ")
    guess = int(guess)
    
    if guess > secret:
        print("Too large")
    elif guess < secret:
        print("Too small")
    else:
        print("You win")
        break
else: # executed in the absence of break
    print(f"You lose: the number was {secret}")


print("Game over")
```
