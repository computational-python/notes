```python
class Person:
    count = 0
    
    def __init__(self, first_name: str, last_name: str):
        self.first_name = first_name
        self.last_name = last_name
        Person.count += 1
        
    def __str__(self): #pretty print
        return self.whoami()  # reuse other function
        
    def __repr__(self): #accurate print
        return f'{self.__class__.__name__}("{self.first_name}", "{self.last_name}")'
        
    def whoami(self):
        return f"My name is {self.last_name}, {self.first_name} {self.last_name}"
```

`first_name:str` is a type annotation

```python
person = Person("John", "Smith")
print(person)
```

```python
repr(person)
```

## Example: complex

```python
class MyComplex:
    def __init__(self, real=0.0, imag=0.0):
        self.real = real
        self.imag = imag

    def __repr__(self):
        return f'{self.__class__.__name__}({self.real}, {self.imag})'

    def __str__(self):
        return f'{self.real} + {self.imag}i'

    def __add__(left, right):
        real_sum = left.real + right.real
        imag_sum = left.imag + right.imag
        return MyComplex(real_sum, imag_sum)
```

```python
MyComplex()
```

```python
print(MyComplex(1, 2))
```

```python
# Addition
z1 = MyComplex(1, 2)
z2 = MyComplex(3, 4)
z1 + z2
```

```python
class Employee(Person):
    
   def __init__(self, first_name, last_name, salary):
       super().__init__(first_name, last_name)
       self.salary = salary

   def __repr__(self): #accurate print
       return f'{self.__class__.__name__}("{self.first_name}", "{self.last_name}", {self.salary})'

   def show_wage(self):
       return f"{self.whoami()}, and I earn {self.salary} kr"
```

```python
emp = Employee('Fred', 'Flintstone', 30000)
```

```python
dir(emp)
```

```python
emp.count
```

```python
emp.whoami()
```
